### Syntax and Semantics in Python 
#### Syntax :
 The set for rules that define the combinations of symbol that considered to be correctly structured programs in a language. ( In simple words syntax is about correct arrangement of words & symbols in a code ) 
#### Semantics: 
It refers to the meaning or the interpretation of the symbols, characters, and commands in a language. It is about what the code is supposed to do when it runs.
