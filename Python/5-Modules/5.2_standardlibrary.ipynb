```python
## Data Serialization
import json
data={"name":"Kulvinder","age":25}

json_str= json.dumps(data)
print(json_str)
print(type(json_str))
past_data=json.loads(json_str)
print(past_data)
print(type(past_data))
```

```python
## CSV
import csv

with open("example.csv", mode="w") as file:
    writer=csv.writer(file)
    writer.writerow(["name","age"])
    writer.writerow(["Kulvinder",25])
 
with open("example.csv", mode="r") as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)

```
