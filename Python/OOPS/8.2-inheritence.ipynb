```python
# Inheritence (Single)
# Parent class 
class Car:
    def __init__(self, windows, doors ,enginetype):

       self.windowss= windows
       self.doors=doors
       self.enginetype=enginetype
    def drive(self):
        print(f"The person will drive {self.enginetype} car") 

car1= Car(4,5,"pertrol")
print(car1.drive())
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, isSelfDriving):
        super().__init__(windows,doors,enginetype,)
        self.isSelfDriving=isSelfDriving
    def selfDriving(self):
        print(f"Tesla supports self driving: {self.isSelfDriving}")
```

```python
tesla=Tesla(4,5,"Electric",True)
print(f"Windows: {tesla.windowss} Doors: {tesla.doors} EngineType: {tesla.enginetype} Is Self Driving? : {tesla.isSelfDriving}")
tesla.selfDriving()
tesla.drive()
```

```python
# Mulity Inheritence
class Animal:
    def __init__(self,name):
        self.name= name 
    def speak(self):
        print("this class wil be implemented in child class")
class Pet:
    def __init__(self,owner):
        self.owner= owner
class Dog(Animal,Pet):
    def __init__(self, name, owner):
        Animal.__init__(self,name) 
        Pet.__init__(self,owner) 
    def speak(self):
        return f"{self.name} says woof"
    # Create object
dog1=Dog("Lucy","Kulvinder")
print(f"Name:{dog1.name} Owner: {dog1.owner}")
dog1.speak()


    
```
