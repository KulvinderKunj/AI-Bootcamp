## Sets
sets are built-in datat type in Python uesd to to store ther collection of unique items.The are unordered, meaning that the elemets not follow a specific order, and they to not allow duplicate elements. Sets are useful for memberships tests , elimenating duplicates entries, and performing mathemetical set operations likr union, intersection, difference,and symmentic difference.  


```python
#creating sets 
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))

```

```python
my_set = set([1,2,3,4,5,6])
print(type(my_set))
print(my_set)
```

```python
my_set = set([1,2,3,4,5,6,1,5,6])
# print(type(my_set))
print(my_set)
```

### Methemetical Operations 

```python
set1 ={1,2,3,4,5,6}
set2= {4,5,6,7,8,9}
# Union 
set_union = set1.union(set2)
print(set_union)
# Intersection 
set_intersection = set1.intersection(set2)
print(set_intersection)
# Intersection 
set1.intersection_update(set2) ## Update the new set in set1
print(set1)
```

Difference 

```python
set1 ={1,2,3,4,5,6}
set2= {4,5,6,7,8,9}

print(set1.difference(set2))
# can use 
# print(set1.difference_update(set2))
print(set2.difference(set1))
```

Symmetric Difference 

```python
print(set1.symmetric_difference(set2))
# unique elements from both sets are combbined  
```
