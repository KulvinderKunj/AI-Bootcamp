## Encapsulation
Encapsulation is the. concept of wrapping data (Variables) and methODS (functions) togeather as a single unit to  restrict direct access to some of the object's components, which means of preventing accidental interference and misuse of the data.

```python
# Encapsulation getter and setter 
# public, protected, and private variable 
class Person:
    def __init__(self, name, age):
        self.name =name #public variable 
        self.age= age #publice variable 

# Public Variable: can use outside the class 



person =Person("Kulvinder", 26)



print(person.age)



def get_name(person):
    return person.name
print(get_name(person))


dir(person)
    
```

```python

class Person2:
    def __init__(self, name, age):
        self.__name =name #private variable 
        self.__age= age #private variable 

# Private Variable: can't use outside the class not even child class 



person =Person2("Kulvinder", 26)




dir(person)
```

```python

class Person3:
    def __init__(self, name, age):
        self._name =name #protected variable 
        self._age= age #protected variable 

# Protected Variable: can't use outside the class but can in child class 

# Child Class 
class Employee(Person3):
    def __init__(self, name, age):
        super().__init__(name, age)


employee= Employee("Kulvinder",26)

print(employee._name)
# print(person.__age)




```
