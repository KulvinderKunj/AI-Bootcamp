```python
import pandas as pd
data= [1,2,3,4,5]
#1D data (1 column data)
series= pd.Series(data)
print("Series: \n" ,series)
print(type(series))

```

```python
# Create series from dictionary

data= {
    'a':1,
    'b':2,
    'c':3
}
series_dic=pd.Series(data)
print(series_dic)
```

```python
# Data Frame  (2D data like table)
# Create dataframe from a dictionry of list
data= {
    "Name":["Kulvinder","Herry","Jack"],
    "Age":[26,24,34],
    'city':["Lahore","Reading","London"]
    
}
df=pd.DataFrame(data)
print(df)
print(type(df))

 
```
