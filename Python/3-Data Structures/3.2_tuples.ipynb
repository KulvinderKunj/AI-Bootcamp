### Immutable Nature of Tuples 
tuples are immutable, meaning their elements cannot be change once assigned

```python
# List is mulable 
lst =[1,2,3,4,5]
print(lst)
lst[0] = "Kulvinder"
print(lst)

#tuple is immutable 
tuple_list= tuple([1,2,3,4,5,6,7])
print(tuple_list) 
# tuple_list[0] = "Singh" #Error


```

Packing an Unpacking

```python
packed_tuples = 1,"Hello", 3.14
print(packed_tuples)
# Unpacking
a,b,c =packed_tuples
print(a,b,c)
```

```python
# Unpacking with *
numbers = (1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)




```

```python
# Nested List
lst = [[1,2,3,4,],[5,6,7,8,9],[1,"hello",3.14, True]]
lst[0][0:3]
```

```python
# Nested Tuples
nested_tuple= ((1,2,3,4,),(5,6,7,8,9),(1,"hello",3.14, True))
nested_tuple[0] 
nested_tuple[1][2] #access 7
```

Iterating Over Nested Tuple 

```python
for sub in nested_tuple:
    for item in sub:
        print(item)

# incomprehension 
lst =[x for sub in nested_tuple for x in sub ]
print(lst)
```
