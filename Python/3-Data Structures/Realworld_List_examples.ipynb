### Example 1 Manage A To Do List 



Create a To DO List To Keep Track of Tasks :

```python
to_do_list =["Buy Groceries","Clean the House","Pay Bills"]

# Adding Tasks
to_do_list.append("Schedule Meeting")

print(to_do_list)
to_do_list.append("Go for a Run")

# Removing Tasks 
to_do_list.remove("Clean the House")
print(to_do_list)


#Checking if a Task is in the list 

if "Pay Bills" in to_do_list:
    print("Please Don't Forget to Pay Bills ")

print("=====TO Do List Remaining=====")
for task in to_do_list:
    print("==>",task)





```

### orgnixing Student Grades 
Create a list to store and calculate avarage Grades for Students 


```python
#orgnizing student grades 
grades =[85,92,90,88]
# adding new grade 
grades.append(95)
# calculating Avarage Grade 
avg_grade = sum(grades)// len(grades)
print("Avarage Grade is :", avg_grade)
# Finding Highest and Lowest Grades 
low_grade = min(grades)
high_grade = max(grades)
print("Lowest Grade is :", low_grade)
print("Highest Grade is :", high_grade)

```

### Example 3 : Managing Inventory 
use a list to manage inventory in a store 


```python
#managing the inventoy 
inventory =["apples","bananas","oranges","grapes"]
# adding inventoris 
inventory.append("strawberries")
#  removing inentory which is out off stock 
inventory.remove("bananas")
# checking if an item is in stock or not 
item ="oranges"
if item in inventory:
    print(f"{item} are in stock.")
else:
    print(f"{item} are out of stock.")

# Printing Invetory 
print("=====Inventory List ======")
for item in inventory:
    print(f"-{item}")
```

### Example:4 Collecting User FeedBack
use a list to collect and analyze the user feed back

```python
# # Collecting the use feedback
feedback = ["Great service!","Very satisfied","Could be better","Excellent experience"]
# adding feedback
feedback.append("Not Happy with the service")
#counting the feedback
positive_feedback_count =sum(1  for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())

print(f"Positive FeedBack Count:{positive_feedback_count} ")
# Printing feedback

print("=====All Feedbacks====")
for comment in feedback:
    print("-",comment)
```
