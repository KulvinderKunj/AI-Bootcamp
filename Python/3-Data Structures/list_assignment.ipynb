### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
def flatten(nested):
    flatten = [item for sublist in nested for item in sublist]
    return flatten


nested_list = [[1,2,3],[3,56,3,8],[13,253,78,1]]

print("Nested List :",nested_list)
print("Flatten List :",flatten(nested_list))
```

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
def Inersection(lst1,lst2):
    return [x for x in list1 if x in list2 ]

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
intersection_list = Inersection(lst1,list2)
print("Intersection:",intersection_list)
```
