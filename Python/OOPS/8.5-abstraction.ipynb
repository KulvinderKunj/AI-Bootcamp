## Abstraction




```python
from abc import ABC ,abstractmethod

class Vehicle(ABC):
    def drive(self):
        print("The Vehicle is ready to Drive")
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        print("Car Engine is Started!")

def start_vehicle(vehicle):
     vehicle.start_engine()
     vehicle.drive()
    



car= Car()
start_vehicle(car)

```
