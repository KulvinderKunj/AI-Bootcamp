```python
# Range 
range(2,5)
```

```python
# for loop (for traversing any collection)
for i in  range(2,5):
    print(i)
```

```python
# for loop and range step
for i in  range(1,10,2):
    print(i)
```

```python
# for loop and range reverse
for i in  range(10,1,-2):
    print(i)
```

```python
# Strings 
str = "Kulvinder"
for i in str:
    print(i)
```

```python
# While loop (repeat Untill the condition is False)

count = 0
while count <5:
    print ("Hello")
    count = count +1
```

```python
# Loop cotrol Statments
# break
# The break statment exits the loop prematurely 
for i in range(10):
    if i == 5:
        break
    else:
        print(i)
```

```python
# Loop cotrol Statments
# continue
# The continue statment skips the current iteration and continues with the next 
for i in range(10):
    if i%2==0:
        print(i)
         
       
       
    else:
        continue
```

```python
# Nested Loops 
#  a loop inside loop
for i in range(3):
    for j in range(2):
        print(f"i = {i}  j = {j}")
```

```python
# Calculate the sum of firt N numbers using while Loop 

n = int (input("Enter the n Number"))
sum = 0

i=1
while i<=n:
    sum = sum+i
    i= i+1



print(f"Sum is {sum}")
```

```python
# Calculate the sum of firt N numbers using while Loop
ng = 10 
sum = 0
count = 1
while count <=ng:
    sum = sum+count
    count = count+1

print(f"Sum is = {sum}")
```
