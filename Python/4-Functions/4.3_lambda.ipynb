### Lambda Function in Python 
Lambda function is a small anonymous functions defined usingh the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to high-order fuctions.

```python
# syntex
lambda arguments: expression 
```

```python
def addition(a,b):
    return a+b
```

```python
addition(2,3)
```

```python
additionl=lambda a,b: a+b

print(additionl(1,2))


```

```python
def even(num):
    if num%2==0:
        return True


print(even(3))
```

```python
even = lambda num1:num1%2==0
print(even(4))
```

```python
def addition (x,y,z):
    return x+y+z
print(addition(1,2,3))
```

```python
adding = lambda x,y,z: x+y+z
print(adding(2,3,4))
```

```python
# Maps map() => it applies a function to all items in a list
#  
number = [1,2,3,4,5,6]

def square(number):
    return number**2


square(2)



# map===> map(function, list )




```

```python
list(map( lambda x: x**2, number))
```
