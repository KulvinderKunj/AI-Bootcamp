```python
# try, except, else
try: 
    num = int(input("Enter a number"))
    result= 10/num
except  ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Please enter the denominator greater than zero")
except Exception as ex:
    print(ex)
else: 
    print(f"The result is {result}")
```
