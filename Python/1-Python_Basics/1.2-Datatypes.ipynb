# ***Data Types***
- Data types are a classsification of data which tell the compiler or interpreter how the programmer intends to use the data.

- They determine the type of opertion that can be preformed on the data, the values that can take, and the amount of the memory meeded to store the data.


## Importance of DataType 
- Data Type ensure that data is stored in an efficent way .
- They help in performing correct operations on data.
- Proper use of Data Types can prevent the errors and bugs in the code.
