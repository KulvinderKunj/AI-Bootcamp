```python
# <!-- Closures functions -->
def main_welcome(msg):
    def sub_welcome():
        print("Welcome from sub function")
        print(msg)
        print("Good Bye!")
    return sub_welcome()
main_welcome("Welcome from Main")
```

```python
# -- Closures functions --
def main_welcome(fun,lst):
    def sub_welcome():
        print("Welcome from sub function")
        print(fun(lst))
        print("Good Bye!")
    return sub_welcome()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
# manual decorators

def main_welcome(fun):
    def sub_welcome():
        print("Welcome from sub function")
        fun()
        print("Good Bye!")
    return sub_welcome()
```

```python
@main_welcome
def course_intro():
    print("Hello welcome to python course")
# course_intro()
```

```python
#Decorator
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is callled. ")
        func()
        print("Something is happening after the function is callled. ")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
# DEcorators with arrguments

def repeat(n):
    def decorator(func):
        def wrapper(*args,**karges):
            for _ in range(n):
                func(*args,**karges)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("hello")


say_hello()
```
