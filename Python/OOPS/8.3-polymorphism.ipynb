## Method Overriding 
Method overriding allows a child class to provinde specific implimentation of a method that is already defined in its parent class

```python
## Base class
class Animal:
    def speak(self):
        return "Sound of the Animal"
# Derived Class 
class Dog(Animal):
    def speak(self):
        return "wooof!"
class Cat(Animal):
    def speak(self):
        return "meow!"
def animal_speak(animal):
    print(animal.speak())
```

```python
dog= Dog()
print(dog.speak())
cat = Cat()
print(cat.speak())
animal=Animal()
print(animal.speak())
print("-----------")
animal_speak(dog)
ami
```
