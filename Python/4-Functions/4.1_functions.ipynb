```python
# Why Functions?
num = 24
if num%2==0:
    print("the Number is Even")
else:
    print("The Number is odd")
```

```python
def even_or_od(num):
    if num%2==0:
     print("the Number is Even")
    else:
      print("The Number is odd")
```

```python
# Default Parameters 
def greet(name ="Sir"):
    print(f" Hello! {name} Welcome to paradise")
greet()
```

```python
#Variable Lenght Argument
#=> Postional 
#=> Keyword Arguments



#================Postional Arguments===============
def print_number(*args ):
    for number in args:
        print(number)

print_number(1,2,3,4,5,6,7,8,9,"Kulvinder") # These Paraperters are called postitional arguments 



```

```python
# ===========Keywords Arguments(All the argumets will be in key value pairs)=======
def print_details(*args,**kwargs):
 
    for val in args:
        print("Postional Arguments:",val)

    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(1,2,3,4,5,6,7,8,9,"Kulvinder",name="Kulvinder",age=25,country="UK")

```
