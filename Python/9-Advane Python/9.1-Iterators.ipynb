## Iterators 


```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
print(my_list)
```

```python
# Iterator 
iterator = iter(my_list)
print(type(iterator))
```

```python
print(iterator)
```

```python
# Iterate through all the elements
try:
    print(next(iterator))
except StopIteration:
   # Iterate through all the elements
    next(iterator)
except StopIteration:
    print("There are no Elemet in the iterator")
```
