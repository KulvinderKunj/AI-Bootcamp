## Generators

```python
def square(num):
    for i in range(num):
        yield i ** 2


    
```

```python
a= square(3)
```

```python
for i in square(5):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen= my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

## Practical Example:
generators are perticularly useful for reading large files. because they allow you to process one line at a time without loading the entire file into memory.

```python
# Reading Large files 

def read_file_read(filename):
    with open(filename, "r") as file:
        for line in file:
            yield line

```

```python
filePath="largefile.txt"


for line in read_file_read(filePath):
    print(line.strip())
```
