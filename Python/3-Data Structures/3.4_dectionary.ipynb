## Dictiories 
Dictionaries are unordered collection of items. They store Data in key-value pairs. Keys must be unique ans immutable (e.g. strings, tuple), while values can be any type.

```python
## Creating Dictionary 
empty_dic = {}
print(type(empty_dic))
```

```python
empty_dic = dict()
empty_dic 
```

```python
student = {
    "name": "Kulvinder",
    "age": 25,
    "grade":"A"
}
print(student)
```

```python
#Error (single key is always used)
student = {
    "name": "Kulvinder",
    "age": 25,
    "grade":"A",
    "name":24

}
print(student)
```

Accessing Dictionary Elements 

```python
student = {
    "name": "Kulvinder",
    "age": 25,
    "grade":"A",
}
print(student)
print(student['grade'])
print(student['age'])
print(student['name'])
```

Another Method of Accessing elements (using get())

```python
print(student.get("grade"))
print(student.get("name"))
print(student.get("age"))

# Last Name (Error)
print(student.get("last_name"))
# default Value
print(student.get("last_name", "Not Availabe"))
```

```python
# Modifying  Dictionary Elements 
#disctioary are mutable, so you can add, update or del elemts 

print(student)

student["age"] =23
print(student)


del student['grade']
print(student)
```

```python
# Dictionaries Methods 
# All Keys 
keys = student.keys()
print(keys)
values = student.values()
print(values)

items = student.items()
print(items)
# print(items)
```

Shallow Copy 

```python
copy_of_student = student
# copy_of_student['age'] =32
print(copy_of_student) 
print(student) ## original data is also updated to solve the we make shelloo coly


# Shellow Copy 


shellow_copy = student.copy()
shellow_copy['age'] =25
print(student)
print(shellow_copy) # Problem Solved 
```

```python
# ITERATING OVER DICTIONARIES 
# You can use loops to iterate over dictionaries , keys , values , items 

#Iterating over keys 
for keys in student.keys():
    print(keys)


```

```python
#Iterating over values
for values in student.values(): 
    print(values)
```

```python
#Iterating over key/value pairs
for key,value in student.items():

    print(f"{key} : {value}")
```

```python
# Nested Diictinaries 
students ={
    "student1": {"name":"Kulvinder", "age":25},
    "student2": {"name":"Emma", "age":23},

    "student3": {"name":"john", "age":27}

    

}
print(students)
```

Accrssing Nested Dictionaries 

```python
print(students['student1']['name'])
print(students['student1']['age'])
```

```python
#Iteraing Over Nested Dictionaries 
for student_id, student_info in students.items():
    # print(f'{student_id}--> Name: {student_info["name"]} Age: {student_info["age"]}')
    print(f'{student_id}:{student_info}')
    for key,value  in student_info.items():
        print(f"{key}: {value}")


```

```python
# Dictionary Comphrehension
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
# condtional iDictionary Comphrehension
even_squres = {x:x**2 for x in range(10) if x%2 ==0}
print(even_squres)
```

##  Practical Examples 

```python
#Use a dictionary to count the frequence of elements in list
numbers = [1,2,2,2,2,3,3,3,4,4,5,5,5,5,5]
frequancy = {}
for number in numbers:
    if number in frequancy: 
        frequancy[number]+=1
    else:
        frequancy[number]=1
print(frequancy)
```

```python
# Merge 2 Dictionries into one
disc1 ={'a':1,'b':2}
disc2 ={'b':3,'c':4}
merged_dict = {**disc1, **disc2}
print(merged_dict)



```
