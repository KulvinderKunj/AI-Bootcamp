### The Map() Function in Python 
The map() function applies a given function to all items of an input list (or any other iterable) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively.


```python
def square(x):
    return x*x
square(4)
```

```python
numbers = [1,2,3,4,5,6,7,8]
list(map(square,numbers))
```

```python
# lambda function with Map()
list(map(lambda x:x*x,numbers))
```

```python
# Map multiple iterable
numbers1=[1,2,3]
numbers2=[4,5,6]
added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
strings= ['1','2','3','4','5']
ints= list(map(int,strings))
print(ints)
print(strings)
```

```python
# In Build Function in Map 
friuts =['apple','banana','cherry']
cap_friuts= list(map(str.upper, friuts))
print(cap_friuts)
```

```python
# Dictionary in Map
def get_name(people):
    return people["name"]
people=[
    {"name": "kulvinder", "age": 25},
    {"name": "Sameer", "age": 24}
]
list(map(get_name,people))
```
